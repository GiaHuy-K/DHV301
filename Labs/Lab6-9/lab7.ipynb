# SETUP

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

df = pd.read_csv("LAB07_student_performance_sample.csv")
df.head()
```

# Định nghĩa target và features

```python
target = "result"
y = df[target].map({"Fail": 0, "Pass": 1})  # encode nhị phân

features_A = ["attendance_rate", "quiz_avg", "assignment_avg", "final_score"]
features_B = ["attendance_rate", "quiz_avg", "assignment_avg"]  # bỏ final_score
```

# Train and test split

```python
def run_baseline(features, label):
    X = df[features]
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=42, stratify=y
    )
    model = LogisticRegression(max_iter=1000)
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    print(f"=== {label} ===")
    print("Accuracy:", accuracy_score(y_test, pred))
    print("Confusion Matrix:\n", confusion_matrix(y_test, pred))
    print(classification_report(y_test, pred, target_names=["Fail","Pass"]))
    return model

model_A = run_baseline(features_A, "Version A (kèm final_score)")
model_B = run_baseline(features_B, "Version B (bỏ final_score)")
```
