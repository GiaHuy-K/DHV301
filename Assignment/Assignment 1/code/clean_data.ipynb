# Part 5: Transformation and preparation

Chỉnh sửa cột lời phê của giáo viên

```python
df['Teacher_Comments'] = df['Teacher_Comments'].str.strip()
df.loc[df['Teacher_Comments'] == '', 'Teacher_Comments'] = 'No comment'
print(f"Số dòng dính khoảng trắng sau xử lý: {(df['Teacher_Comments'] == '').sum()}")
```

Chỉnh sửa cột Subject

```python
df['Subject'] = df['Subject'].astype('category')
print("Kiểu dữ liệu mới của cột Subject:", df['Subject'].dtype)
```

Chỉnh sửa cột Homework_Completion_%

```python
df['Homework_Completion_%'] = df['Homework_Completion_%'].astype(str).str.replace('%', '', regex=False)
df['Homework_Completion_%'] = pd.to_numeric(df['Homework_Completion_%'], errors='coerce')

valid_homework_median = df[df['Homework_Completion_%'] >= 0]['Homework_Completion_%'].median()
print(f"Giá trị trung vị hợp lệ tính được: {valid_homework_median}%")

df.loc[df['Homework_Completion_%'] < 0, 'Homework_Completion_%'] = valid_homework_median

print("\nTần suất các giá trị sau khi làm sạch:")
print(df['Homework_Completion_%'].value_counts())
```

Chỉnh sửa cột Exam_Score

```python
df['Exam_Score'] = df['Exam_Score'].clip(upper=100)
```

Giải quyết việc trùng ID và subject


```python
# Sắp xếp giảm dần theo điểm thi để dòng điểm cao nhất lên đầu
df = df.sort_values(by=['Student_ID', 'Subject', 'Exam_Score'], ascending=[True, True, False])

# Chỉ giữ lại bản ghi điểm cao nhất cho mỗi cặp (Student_ID, Subject)
df_cleaned = df.drop_duplicates(subset=['Student_ID', 'Subject'], keep='first').copy()

df_cleaned = df_cleaned.reset_index(drop=True)
df_cleaned
```

```python
df_cleaned['Exam_Status'] = np.where(df_cleaned['Exam_Score'] >= 50, 'Pass', 'Fail')

# Academic_Grade
def classify_grade(score):
    if score >= 90: return 'Excellent'
    elif score >= 80: return 'Good'
    elif score >= 65: return 'Average'
    elif score >= 50: return 'Below Average'
    else: return 'Weak'

df_cleaned['Academic_Grade'] = df_cleaned['Exam_Score'].apply(classify_grade)

# Nhóm mức độ hoàn thành bài tập (Homework_Effort_Group)
df_cleaned['Homework_Effort_Group'] = pd.cut(
    df_cleaned['Homework_Completion_%'],
    bins=[-1, 85, 95, 101],
    labels=['Low Effort', 'Medium Effort', 'High Effort']
)

# Cờ cảnh báo rủi ro học tập sớm (Academic_Risk_Flag)
# Gắn cờ 1 (Nguy cơ cao) nếu điểm thi < 50 VÀ tỷ lệ bài tập về nhà ở mức thấp (<= 85)
df_cleaned['Academic_Risk_Flag'] = np.where(
    (df_cleaned['Exam_Score'] < 50) & (df_cleaned['Homework_Completion_%'] <= 85), 
    1, 0
)
df_cleaned
```

Check lại sau khi sửa đổi

```python
print(f"Kích thước cuối cùng của Dataset: {df_cleaned.shape[0]} dòng và {df_cleaned.shape[1]} cột.")
print(f"- Số lượng ô trống NaN: {df_cleaned.isnull().sum().sum()}")
print(f"- Số dòng dính giá trị điểm > 100: {len(df_cleaned[df_cleaned['Exam_Score'] > 100])}")
print(f"- Số dòng dính giá trị bài tập âm (<0): {len(df_cleaned[df_cleaned['Homework_Completion_%'] < 0])}")
print(f"- Số dòng trùng lặp logic còn sót lại: {df_cleaned.duplicated(subset=['Student_ID', 'Subject']).sum()}")
print(f"- Số dòng dính lỗi văn bản trống : {(df_cleaned['Teacher_Comments'].str.strip() == '').sum()}")
```

```python

output_file_path = os.path.join(base_path, '..', 'data_cleaned', 'performance_cleaned.csv')

# Lấy đường dẫn thư mục cha để kiểm tra và tự động tạo nếu chưa có
output_dir = os.path.dirname(output_file_path)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"Đã tạo thư mục mới: {output_dir}")

# Lưu dữ liệu sạch ra file
df_cleaned.to_csv(output_file_path, index=False, encoding='utf-8-sig')

print(f"🎉 Đã lưu dữ liệu sạch thành công tại: {output_file_path}")
```
